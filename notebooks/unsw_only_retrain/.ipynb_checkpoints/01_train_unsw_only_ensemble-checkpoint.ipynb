```python
import os, json, warnings, time
import numpy as np
import pandas as pd
import joblib
from datetime import datetime

os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"
os.environ["TF_ENABLE_ONEDNN_OPTS"] = "0"
warnings.filterwarnings("ignore")

ROOT = "C:/Bastion_IDS"
DATA = os.path.join(ROOT, "data")
OUT = os.path.join(ROOT, "notebooks", "unsw_only_retrain", "outputs")
MODEL_OUT = os.path.join(OUT, "models")
RESULTS_OUT = os.path.join(OUT, "results")
os.makedirs(MODEL_OUT, exist_ok=True)
os.makedirs(RESULTS_OUT, exist_ok=True)

from sklearn.preprocessing import LabelEncoder, StandardScaler, OrdinalEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.model_selection import StratifiedShuffleSplit, RandomizedSearchCV
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (classification_report, accuracy_score, matthews_corrcoef,
                             roc_auc_score, f1_score, precision_score, recall_score)
from sklearn.utils.class_weight import compute_class_weight
from xgboost import XGBClassifier
from catboost import CatBoostClassifier
from imblearn.over_sampling import BorderlineSMOTE
from imblearn.under_sampling import EditedNearestNeighbours
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, callbacks
```

```python
def log(msg):
    print(f"[{datetime.now().strftime('%H:%M:%S')}] {msg}", flush=True)

CAT_FEATURES = ["proto", "service", "state"]
NUM_FEATURES = [
    "dur","spkts","dpkts","sbytes","dbytes","rate","sttl","dttl",
    "sload","dload","sloss","dloss","sinpkt","dinpkt","sjit","djit",
    "swin","stcpb","dtcpb","dwin","tcprtt","synack","ackdat","smean",
    "dmean","trans_depth","response_body_len","ct_srv_src","ct_state_ttl",
    "ct_dst_ltm","ct_src_dport_ltm","ct_dst_sport_ltm","ct_dst_src_ltm",
    "is_ftp_login","ct_ftp_cmd","ct_flw_http_mthd","ct_src_ltm",
    "ct_srv_dst","is_sm_ips_ports",
]
ATTACK_CLASSES = ['Analysis','Backdoor','DoS','Exploits','Fuzzers',
                  'Generic','Normal','Reconnaissance','Shellcode','Worms']

def compute_metrics(y_true, y_pred, y_prob, classes):
    report = classification_report(y_true, y_pred, target_names=classes,
                                   zero_division=0, output_dict=True)
    mcc = matthews_corrcoef(y_true, y_pred)
    acc = accuracy_score(y_true, y_pred)
    f1m = f1_score(y_true, y_pred, average="macro", zero_division=0)
    f1w = f1_score(y_true, y_pred, average="weighted", zero_division=0)
    rec = recall_score(y_true, y_pred, average="macro", zero_division=0)
    pre = precision_score(y_true, y_pred, average="macro", zero_division=0)
    try:
        auc = roc_auc_score(pd.get_dummies(y_true), y_prob, multi_class="ovr", average="macro")
    except Exception:
        auc = float("nan")
    return {
        "accuracy": round(acc,4), "f1_macro": round(f1m,4), "f1_weighted": round(f1w,4),
        "recall_macro": round(rec,4), "precision_macro": round(pre,4), "mcc": round(mcc,4),
        "auc_roc_macro": round(auc,4), "classification_report": report,
    }
```

```python
# ------------------------------------------------------------
# 1. LOAD UNSW-NB15 (train + test)
# ------------------------------------------------------------
log("Loading UNSW-NB15...")
df_train = pd.read_csv(os.path.join(DATA, "UNSW_NB15_training-set.csv"))
df_test = pd.read_csv(os.path.join(DATA, "UNSW_NB15_testing-set.csv"))
df_unsw = pd.concat([df_train, df_test], ignore_index=True)
df_unsw["attack_cat"] = df_unsw["attack_cat"].str.strip().fillna("Normal")
df_unsw = df_unsw.drop(columns=["id","label"], errors="ignore")
log(f"UNSW combined: {len(df_unsw):,} rows")

feature_cols = CAT_FEATURES + NUM_FEATURES
df_all = df_unsw[feature_cols + ["attack_cat"]].copy()
df_all[CAT_FEATURES] = df_all[CAT_FEATURES].fillna("-").astype(str).apply(lambda c: c.str.lower().str.strip())
df_all[NUM_FEATURES] = df_all[NUM_FEATURES].apply(pd.to_numeric, errors="coerce").fillna(0)
df_all[NUM_FEATURES] = df_all[NUM_FEATURES].replace([np.inf, -np.inf], 0)
df_all["attack_cat"] = df_all["attack_cat"].str.strip()
df_all = df_all[df_all["attack_cat"].isin(ATTACK_CLASSES)].reset_index(drop=True)
log(f"Training dataset: {len(df_all):,} rows")
```

```python
# ------------------------------------------------------------
# 2. ENCODE LABELS, FIT PREPROCESSOR, SPLIT
# ------------------------------------------------------------
le = LabelEncoder()
le.fit(ATTACK_CLASSES)
y = le.transform(df_all["attack_cat"])
X_raw = df_all[feature_cols]

cat_pipe = Pipeline([("enc", OrdinalEncoder(handle_unknown="use_encoded_value", unknown_value=-1))])
num_pipe = Pipeline([("scaler", StandardScaler())])
preprocessor = ColumnTransformer([
    ("cat", cat_pipe, CAT_FEATURES),
    ("num", num_pipe, NUM_FEATURES),
], remainder="drop")

sss = StratifiedShuffleSplit(n_splits=1, test_size=0.20, random_state=42)
train_idx, test_idx = next(sss.split(X_raw, y))
X_train_raw, X_test_raw = X_raw.iloc[train_idx], X_raw.iloc[test_idx]
y_train, y_test = y[train_idx], y[test_idx]

preprocessor.fit(X_train_raw)
X_train = preprocessor.transform(X_train_raw)
X_test = preprocessor.transform(X_test_raw)
n_features = X_train.shape[1]
log(f"Train: {X_train.shape}, Test: {X_test.shape}")

joblib.dump(preprocessor, os.path.join(OUT, "preprocessor.pkl"))
joblib.dump(le, os.path.join(OUT, "label_encoder.pkl"))
```

```python
# ------------------------------------------------------------
# 3. REBALANCE TRAINING DATA -- BORDERLINE-SMOTE + ENN (training split only)
# ------------------------------------------------------------
counts = np.bincount(y_train, minlength=len(le.classes_))
strategy = {i: max(counts[i], 800) for i in range(len(le.classes_)) if counts[i] < 800}
if strategy:
    bsmote = BorderlineSMOTE(sampling_strategy=strategy, random_state=42, k_neighbors=3)
    X_train_over, y_train_over = bsmote.fit_resample(X_train, y_train)
else:
    X_train_over, y_train_over = X_train, y_train
log(f"Post-Borderline-SMOTE train size: {X_train_over.shape[0]:,}")

enn = EditedNearestNeighbours(n_jobs=-1, sampling_strategy="majority")
X_train_sm, y_train_sm = enn.fit_resample(X_train_over, y_train_over)
log(f"Post-ENN train size: {X_train_sm.shape[0]:,} (majority class only, minority classes untouched)")

all_metrics = {}
```

```python
# ------------------------------------------------------------
# 4. RANDOM FOREST (CPU, no GPU backend in sklearn)
# ------------------------------------------------------------
log("Training Random Forest...")
t0 = time.time()
rf_params = {
    "n_estimators":    [200, 300, 400],
    "max_depth":       [20, 30, None],
    "min_samples_split": [2, 4],
    "min_samples_leaf":  [1, 2],
    "max_features":    ["sqrt", "log2"],
}
rf_base = RandomForestClassifier(class_weight="balanced", n_jobs=-1, random_state=42)
rf_search = RandomizedSearchCV(rf_base, rf_params, n_iter=12, cv=3,
                               scoring="f1_macro", n_jobs=-1,
                               random_state=42, verbose=1)
sub_idx = np.random.RandomState(42).choice(len(X_train_sm), min(50000, len(X_train_sm)), replace=False)
rf_search.fit(X_train_sm[sub_idx], y_train_sm[sub_idx])
log(f"Best RF params: {rf_search.best_params_}")

rf_final = RandomForestClassifier(**rf_search.best_params_,
                                  class_weight="balanced", n_jobs=-1, random_state=42)
rf_final.fit(X_train_sm, y_train_sm)
log(f"RF trained in {time.time()-t0:.0f}s")

y_pred_rf = rf_final.predict(X_test)
y_prob_rf = rf_final.predict_proba(X_test)
m_rf = compute_metrics(y_test, y_pred_rf, y_prob_rf, list(le.classes_))
all_metrics["random_forest"] = m_rf
log(f"RF -- acc {m_rf['accuracy']:.4f}, f1_macro {m_rf['f1_macro']:.4f}")

joblib.dump(rf_final, os.path.join(MODEL_OUT, "randomforest.pkl"))
```

```python
# ------------------------------------------------------------
# 5. XGBOOST (GPU)
# ------------------------------------------------------------
log("Training XGBoost...")
t0 = time.time()
xgb_params = {
    "n_estimators":    [200, 300, 400],
    "max_depth":       [4, 6, 8],
    "learning_rate":   [0.05, 0.1, 0.15],
    "subsample":       [0.8, 1.0],
    "colsample_bytree":[0.8, 1.0],
    "min_child_weight":[1, 3],
    "gamma":           [0, 0.05, 0.1],
}
xgb_base = XGBClassifier(objective="multi:softprob",
                          num_class=len(le.classes_),
                          eval_metric="mlogloss",
                          tree_method="hist",
                          device="cuda",
                          random_state=42)
xgb_search = RandomizedSearchCV(xgb_base, xgb_params, n_iter=15, cv=3,
                                scoring="f1_macro", n_jobs=1,
                                random_state=42, verbose=1)
xgb_search.fit(X_train_sm[sub_idx], y_train_sm[sub_idx])
log(f"Best XGB params: {xgb_search.best_params_}")

xgb_final = XGBClassifier(**xgb_search.best_params_,
                           objective="multi:softprob",
                           num_class=len(le.classes_),
                           eval_metric="mlogloss",
                           tree_method="hist",
                           device="cuda",
                           random_state=42)
xgb_final.fit(X_train_sm, y_train_sm, eval_set=[(X_test, y_test)], verbose=False)
log(f"XGB trained in {time.time()-t0:.0f}s")

y_pred_xgb = xgb_final.predict(X_test)
y_prob_xgb = xgb_final.predict_proba(X_test)
m_xgb = compute_metrics(y_test, y_pred_xgb, y_prob_xgb, list(le.classes_))
all_metrics["xgboost"] = m_xgb
log(f"XGB -- acc {m_xgb['accuracy']:.4f}, f1_macro {m_xgb['f1_macro']:.4f}")

xgb_final.save_model(os.path.join(MODEL_OUT, "xgboost.json"))
joblib.dump(xgb_final, os.path.join(MODEL_OUT, "xgboost.pkl"))
```

```python
# ------------------------------------------------------------
# 6. CATBOOST (GPU)
# ------------------------------------------------------------
log("Training CatBoost...")
t0 = time.time()
cat_final = CatBoostClassifier(
    iterations=500, depth=8, learning_rate=0.1,
    loss_function="MultiClass", eval_metric="Accuracy",
    random_seed=42, verbose=50,
    early_stopping_rounds=30,
    auto_class_weights="Balanced",
    task_type="GPU", devices="0",
)
cat_final.fit(X_train_sm, y_train_sm, eval_set=(X_test, y_test), use_best_model=True)
log(f"CatBoost trained in {time.time()-t0:.0f}s")

y_pred_cat = cat_final.predict(X_test).flatten()
y_prob_cat = cat_final.predict_proba(X_test)
m_cat = compute_metrics(y_test, y_pred_cat, y_prob_cat, list(le.classes_))
all_metrics["catboost"] = m_cat
log(f"CatBoost -- acc {m_cat['accuracy']:.4f}, f1_macro {m_cat['f1_macro']:.4f}")

joblib.dump(cat_final, os.path.join(MODEL_OUT, "catboost.pkl"))
```

```python
# ------------------------------------------------------------
# 7. DNN SPECIALIST -- residual MLP (CPU, TF has no native GPU support on Windows)
# ------------------------------------------------------------
log("Training DNN specialist...")
t0 = time.time()
n_classes = len(le.classes_)

def build_dnn(n_feat, n_cls):
    inp = keras.Input(shape=(n_feat,))
    x = layers.Dense(256, activation="relu")(inp)
    x = layers.BatchNormalization()(x)
    x = layers.Dropout(0.3)(x)
    x = layers.Dense(256, activation="relu")(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dropout(0.3)(x)
    skip = x
    x = layers.Dense(128, activation="relu")(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dropout(0.25)(x)
    x = layers.Dense(256, activation="relu")(x)
    x = layers.Add()([x, skip])
    x = layers.Dense(128, activation="relu")(x)
    x = layers.Dropout(0.2)(x)
    x = layers.Dense(64, activation="relu")(x)
    out = layers.Dense(n_cls, activation="softmax")(x)
    return keras.Model(inp, out)

cw = compute_class_weight("balanced", classes=np.unique(y_train_sm), y=y_train_sm)
cw_dict = dict(enumerate(cw))

dnn = build_dnn(n_features, n_classes)
dnn.compile(optimizer=keras.optimizers.Adam(learning_rate=0.001),
            loss="sparse_categorical_crossentropy", metrics=["accuracy"])

dnn_checkpoint_path = os.path.join(MODEL_OUT, "dnn_specialist.keras")
cb_list = [
    callbacks.EarlyStopping(patience=8, restore_best_weights=True, monitor="val_accuracy"),
    callbacks.ReduceLROnPlateau(factor=0.5, patience=4, min_lr=1e-5),
    callbacks.ModelCheckpoint(dnn_checkpoint_path, save_best_only=True, monitor="val_accuracy"),
]

dnn.fit(X_train_sm, y_train_sm,
        validation_data=(X_test, y_test),
        epochs=60, batch_size=512,
        class_weight=cw_dict,
        callbacks=cb_list, verbose=1)
log(f"DNN trained in {time.time()-t0:.0f}s")

dnn = keras.models.load_model(dnn_checkpoint_path)
y_pred_dnn_prob = dnn.predict(X_test, batch_size=1024, verbose=0)
y_pred_dnn = np.argmax(y_pred_dnn_prob, axis=1)
m_dnn = compute_metrics(y_test, y_pred_dnn, y_pred_dnn_prob, list(le.classes_))
all_metrics["dnn"] = m_dnn
log(f"DNN -- acc {m_dnn['accuracy']:.4f}, f1_macro {m_dnn['f1_macro']:.4f}")
```

```python
# ------------------------------------------------------------
# 8. ENSEMBLE FUSION METRICS (simple probability average)
# ------------------------------------------------------------
prob_ensemble = (y_prob_rf + y_prob_xgb + y_prob_cat + y_pred_dnn_prob) / 4
y_pred_ens = np.argmax(prob_ensemble, axis=1)
m_ens = compute_metrics(y_test, y_pred_ens, prob_ensemble, list(le.classes_))
all_metrics["hybrid_ensemble"] = m_ens
log(f"Ensemble -- acc {m_ens['accuracy']:.4f}, f1_macro {m_ens['f1_macro']:.4f}")
```

```python
# ------------------------------------------------------------
# 9. SAVE METRICS REPORT
# ------------------------------------------------------------
report_path = os.path.join(RESULTS_OUT, "metrics_report.json")
report = {
    "generated_at": datetime.now().isoformat(),
    "dataset": "UNSW-NB15 only (no CICIDS)",
    "total_samples": len(df_all),
    "train_samples": len(X_train_sm),
    "test_samples": len(X_test),
    "attack_classes": ATTACK_CLASSES,
    "models": all_metrics,
}
with open(report_path, "w") as f:
    json.dump(report, f, indent=2)

log(f"Metrics report -> {report_path}")
for name, m in all_metrics.items():
    log(f"  {name:<16} acc={m['accuracy']:.4f}  f1_macro={m['f1_macro']:.4f}  "
        f"rec={m['recall_macro']:.4f}  mcc={m['mcc']:.4f}  auc={m['auc_roc_macro']:.4f}")
```
