# Bastion IDS — Systems-Level Evaluation

Measures throughput, per-flow latency, and memory footprint of the running backend.

**Requirements before running:**
- Bastion backend must be running (dev mode: `cd app-desktop && npm run start`, or the packaged app)
- This notebook must run in the Bastion venv kernel
- No active live-capture session during measurement (capture thread adds noise)

**What this measures:**
- Per-flow analysis latency (p50 / p95 / p99) through the full 4-layer pipeline
- Sustained throughput (flows per second)
- Backend process RSS memory at idle, under load, and at peak

Results are saved to `outputs/systems_eval_results.json`.

```python
import os, json, time, statistics, threading
import requests
import psutil
import pandas as pd
import numpy as np

BASE_URL    = "http://127.0.0.1:48217"
AUTH        = "BASTION-KADIAN-SEC-0x42"
TEST_CSV    = r"C:\Bastion_IDS\data\UNSW_NB15_testing-set.csv"
OUT_DIR     = os.path.join(os.path.dirname(os.path.abspath('__file__')), "outputs")
OUT_DIR     = r"C:\Bastion_IDS\notebooks\systems_eval\outputs"
os.makedirs(OUT_DIR, exist_ok=True)

N_WARMUP  = 20    # requests to discard (JIT / model-cache warming)
N_MEASURE = 300   # requests to time (mix of attack + normal rows)

HEADERS = {
    "x-authority": AUTH,
    "Content-Type": "application/json"
}

# Features the model expects (42 UNSW-NB15 columns)
UNSW_FEATURES = [
    "dur","proto","service","state","spkts","dpkts","sbytes","dbytes",
    "rate","sttl","dttl","sload","dload","sloss","dloss","sinpkt","dinpkt",
    "sjit","djit","swin","stcpb","dtcpb","dwin","tcprtt","synack","ackdat",
    "smean","dmean","trans_depth","response_body_len","ct_srv_src","ct_state_ttl",
    "ct_dst_ltm","ct_src_dport_ltm","ct_dst_sport_ltm","ct_dst_src_ltm",
    "is_ftp_login","ct_ftp_cmd","ct_flw_http_mthd","ct_src_ltm","ct_srv_dst",
    "is_sm_ips_ports"
]

print("Configuration OK.")
```

```python
# ── 1. Verify backend is reachable ────────────────────────────────────────────
try:
    r = requests.get(f"{BASE_URL}/api/v1/health", headers=HEADERS, timeout=5)
    r.raise_for_status()
    print(f"Backend healthy: {r.json()}")
except Exception as e:
    raise RuntimeError(
        f"Cannot reach backend at {BASE_URL}. "
        "Start Bastion (dev mode or packaged app) before running this notebook.\n"
        f"Error: {e}"
    )
```

```python
# ── 2. Find backend process for memory sampling ───────────────────────────────
backend_proc = None
for conn in psutil.net_connections(kind="inet"):
    if conn.laddr.port == 48217 and conn.status == "LISTEN":
        try:
            backend_proc = psutil.Process(conn.pid)
            break
        except Exception:
            pass

if backend_proc is None:
    print("WARNING: could not find backend process on port 48217 via psutil.")
    print("Memory measurement will be skipped.")
else:
    idle_rss_mb = backend_proc.memory_info().rss / 1_048_576
    print(f"Backend process PID: {backend_proc.pid}")
    print(f"Idle RSS: {idle_rss_mb:.1f} MB")
```

```python
# ── 3. Load test flows from UNSW-NB15 test set ───────────────────────────────
df_full = pd.read_csv(TEST_CSV, low_memory=False)
# Keep only the feature columns (drop id, attack_cat, label)
df_flows = df_full[UNSW_FEATURES].copy()

# Sample: balanced attack/normal mix (realistic traffic distribution)
np.random.seed(99)
n_needed = N_WARMUP + N_MEASURE
# ~60% normal, 40% attack — matches UNSW test set distribution
n_normal = int(n_needed * 0.60)
n_attack = n_needed - n_normal
idx_normal = df_full[df_full["label"] == 0].index
idx_attack = df_full[df_full["label"] == 1].index
sample_idx = np.concatenate([
    np.random.choice(idx_normal, n_normal, replace=False),
    np.random.choice(idx_attack, n_attack, replace=False)
])
np.random.shuffle(sample_idx)
flow_dicts = df_flows.loc[sample_idx].to_dict(orient="records")

print(f"Loaded {len(flow_dicts)} flow vectors ({n_normal} normal, {n_attack} attack).")
print(f"Warmup: {N_WARMUP} | Measurement: {N_MEASURE}")
```

```python
# ── 4. Warmup pass (not counted) ─────────────────────────────────────────────
print("Warming up...")
for i, fd in enumerate(flow_dicts[:N_WARMUP]):
    requests.post(
        f"{BASE_URL}/api/v1/analyze",
        headers=HEADERS,
        json={"flow": {k: (v if pd.notna(v) else 0) for k, v in fd.items()}},
        timeout=30
    )
print(f"Warmup done ({N_WARMUP} requests discarded).")
```

```python
# ── 5. Measurement pass ───────────────────────────────────────────────────────
# Memory sampler runs in a background thread, sampling every 0.5 s
rss_samples = []
stop_sampler = threading.Event()

def _sample_memory():
    while not stop_sampler.is_set():
        if backend_proc is not None:
            try:
                rss_samples.append(backend_proc.memory_info().rss / 1_048_576)
            except Exception:
                pass
        stop_sampler.wait(timeout=0.5)

sampler = threading.Thread(target=_sample_memory, daemon=True)
sampler.start()

latencies_ms = []
verdicts = []
errors = 0

measure_flows = flow_dicts[N_WARMUP : N_WARMUP + N_MEASURE]
wall_start = time.perf_counter()

for i, fd in enumerate(measure_flows):
    clean = {k: (v if pd.notna(v) else 0) for k, v in fd.items()}
    t0 = time.perf_counter()
    try:
        resp = requests.post(
            f"{BASE_URL}/api/v1/analyze",
            headers=HEADERS,
            json={"flow": clean},
            timeout=30
        )
        t1 = time.perf_counter()
        if resp.status_code == 200:
            latencies_ms.append((t1 - t0) * 1000)
            verdict = resp.json().get("verdict", "?")
            verdicts.append(verdict)
        else:
            errors += 1
    except Exception:
        errors += 1
    if (i + 1) % 50 == 0:
        print(f"  {i+1}/{N_MEASURE} flows done...")

wall_elapsed = time.perf_counter() - wall_start
stop_sampler.set()
sampler.join(timeout=2)

print(f"\nMeasurement done. {len(latencies_ms)} successful, {errors} errors.")
```

```python
# ── 6. Compute statistics ─────────────────────────────────────────────────────
lat = sorted(latencies_ms)
n   = len(lat)

def pct(data, p):
    idx = max(0, int(len(data) * p / 100) - 1)
    return data[idx]

p50  = statistics.median(lat)
p95  = pct(lat, 95)
p99  = pct(lat, 99)
mean = statistics.mean(lat)
throughput = n / wall_elapsed   # flows per second (sequential)

if rss_samples:
    load_rss_peak = max(rss_samples)
    load_rss_mean = statistics.mean(rss_samples)
else:
    load_rss_peak = load_rss_mean = None

print("\n══ BASTION IDS SYSTEMS-LEVEL EVALUATION ══")
print(f"  Flows measured           : {n}")
print(f"  Errors                   : {errors}")
print(f"")
print(f"  Latency (ms)")
print(f"    Mean                   : {mean:7.1f} ms")
print(f"    Median (p50)           : {p50:7.1f} ms")
print(f"    p95                    : {p95:7.1f} ms")
print(f"    p99                    : {p99:7.1f} ms")
print(f"")
print(f"  Throughput (sequential)  : {throughput:7.1f} flows/sec")
print(f"  Total wall time          : {wall_elapsed:7.1f} s for {n} flows")
print(f"")
if backend_proc is not None:
    print(f"  Memory (RSS)")
    print(f"    Idle                 : {idle_rss_mb:7.1f} MB")
    if load_rss_peak is not None:
        print(f"    Under load (peak)    : {load_rss_peak:7.1f} MB")
        print(f"    Under load (mean)    : {load_rss_mean:7.1f} MB")
print()
```

```python
# ── 7. Save results JSON ──────────────────────────────────────────────────────
out = {
    "n_measured":        n,
    "n_errors":          errors,
    "latency_ms": {
        "mean":          round(mean, 2),
        "p50":           round(p50, 2),
        "p95":           round(p95, 2),
        "p99":           round(p99, 2),
        "min":           round(lat[0], 2),
        "max":           round(lat[-1], 2),
    },
    "throughput_flows_per_sec": round(throughput, 2),
    "wall_elapsed_sec":         round(wall_elapsed, 2),
    "memory_mb": {
        "idle":          round(idle_rss_mb, 1) if backend_proc is not None else None,
        "load_peak":     round(load_rss_peak, 1) if load_rss_peak else None,
        "load_mean":     round(load_rss_mean, 1) if load_rss_mean else None,
    },
    "full_latencies_ms":        [round(x, 3) for x in latencies_ms],
}

out_path = os.path.join(OUT_DIR, "systems_eval_results.json")
with open(out_path, "w") as f:
    json.dump(out, f, indent=2)

print(f"Results saved to: {out_path}")
```
